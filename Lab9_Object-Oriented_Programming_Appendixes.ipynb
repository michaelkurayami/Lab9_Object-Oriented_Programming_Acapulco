```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.available = True

    def borrow(self):
        if self.available:
            self.available = False
            print("Book successfully borrowed.")
        else:
            print("Book is already borrowed.")

    def return_book(self):
        if not self.available:
            self.available = True
            print("Book successfully returned.")
        else:
            print("Book is already available.")

class Library:
    def __init__(self):
        self.books = []

    def add_book(self, title, author):
        self.books.append(Book(title, author))

    def show_books(self):
        for book in self.books:
            status = "Available" if book.available else "Borrowed"
            print(book.title, "-", book.author, "-", status)

    def find_book(self, title):
        for book in self.books:
            if book.title.lower() == title.lower():
                return book
        return None

library = Library()

number = int(input("Enter number of books to add: "))

for i in range(number):
    title = input("Enter book title: ")
    author = input("Enter author: ")
    library.add_book(title, author)

while True:
    print("\n1. Borrow Book")
    print("2. Return Book")
    print("3. Display Books")
    print("4. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        title = input("Enter book title: ")
        book = library.find_book(title)

        if book:
            book.borrow()
        else:
            print("Book does not exist.")

    elif choice == "2":
        title = input("Enter book title: ")
        book = library.find_book(title)

        if book:
            book.return_book()
        else:
            print("Book does not exist.")

    elif choice == "3":
        library.show_books()

    elif choice == "4":
        break

    else:
        print("Invalid choice.")
```

```python
class Employee:
    def __init__(self, name, hourly_rate, hours_worked):
        self.name = name
        self.hourly_rate = hourly_rate
        self.hours_worked = hours_worked

    def calculate_salary(self):
        regular_hours = min(self.hours_worked, 160)
        overtime_hours = max(self.hours_worked - 160, 0)

        base_salary = regular_hours * self.hourly_rate
        overtime_pay = overtime_hours * self.hourly_rate * 1.5
        total_salary = base_salary + overtime_pay

        return base_salary, overtime_pay, total_salary

employees = []

number = int(input("Enter number of employees: "))

if number < 1:
    print("At least one employee is required.")

else:
    for i in range(number):
        print("\nEmployee", i + 1)

        name = input("Enter employee name: ")
        rate = float(input("Enter hourly rate: "))
        hours = float(input("Enter hours worked: "))

        if rate < 0 or hours < 0:
            print("Invalid input. Values cannot be negative.")
            continue

        employees.append(Employee(name, rate, hours))

    for employee in employees:
        base, overtime, total = employee.calculate_salary()

        print("\nEmployee:", employee.name)
        print("Total Hours:", employee.hours_worked)
        print("Base Salary: ₱", base)
        print("Overtime Pay: ₱", overtime)
        print("Total Salary: ₱", total)
```

```python
class Product:
    def __init__(self, name, price, stock):
        self.name = name
        self.price = price
        self.stock = stock

products = {
    "1": Product("Keyboard", 500, 10),
    "2": Product("Mouse", 250, 15),
    "3": Product("Monitor", 5000, 5)
}

cart = {}

def show_products():
    print("\nAvailable Products:")
    for key, product in products.items():
        print(key, product.name, "- ₱", product.price,
              "- Stock:", product.stock)

def show_cart():
    total = 0
    print("\nShopping Cart:")

    for key, quantity in cart.items():
        product = products[key]
        subtotal = product.price * quantity
        total += subtotal

        print(product.name, "x", quantity, "- ₱", subtotal)

    print("Total Price: ₱", total)

while True:
    print("\n1. View Products")
    print("2. Add Product")
    print("3. Remove Product")
    print("4. View Cart")
    print("5. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        show_products()

    elif choice == "2":
        show_products()
        key = input("Enter product number: ")

        if key not in products:
            print("Invalid product selection.")
            continue

        quantity = int(input("Enter quantity: "))
        product = products[key]

        if quantity <= 0:
            print("Invalid quantity.")
        elif quantity > product.stock:
            print("Not enough stock available.")
        else:
            cart[key] = cart.get(key, 0) + quantity
            product.stock -= quantity
            print("Product successfully added.")

    elif choice == "3":
        key = input("Enter product number to remove: ")

        if key not in cart:
            print("Product is not in the cart.")
        else:
            quantity = int(input("Enter quantity to remove: "))

            if quantity <= 0 or quantity > cart[key]:
                print("Invalid quantity.")
            else:
                cart[key] -= quantity
                products[key].stock += quantity

                if cart[key] == 0:
                    del cart[key]

                print("Product successfully removed.")

    elif choice == "4":
        show_cart()

    elif choice == "5":
        print("Thank you for shopping!")
        break

    else:
        print("Invalid choice.")
```
